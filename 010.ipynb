$\text{Just messing around with 3D plots}$

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
_ = np.linspace(-2, 2, 100)
x, y = np.meshgrid(_, _)
f = np.exp(-(x**2 + y**2))*np.sin(x)
fig, ax =plt.subplots(subplot_kw={'projection':'3d'}, figsize=(6,6))
ax.plot_surface(x, y, f, cmap='plasma')
ax.set_axis_off() 
ax.set_title('3D Plot of $f(x, y) = e^{-(x^2 + y^2)}\\sin(x)$', fontsize=14)
```

```python
_ = np.linspace(-3, 3, 100)
x, y = np.meshgrid(_, _)
f = np.exp(-(x**2 + y**2)) * (1 - (x**2 + y**2))
fig, ax =plt.subplots(subplot_kw={'projection':'3d'}, figsize=(6,6))
ax.plot_surface(x, y, f, cmap='plasma')
ax.set_axis_off() 
ax.set_title('3D Plot of $f(x, y) = e^{-(x^2 + y^2)}(1 - x^2 - y^2)$', fontsize=14)
```

```python
_ = np.linspace(-2, 2, 100)
x, y = np.meshgrid(_, _)
f = np.exp(-(x**2 + y**2)) * np.sin(2*x) * np.cos(2*y)
fig, axes =plt.subplots(1, 2,subplot_kw={'projection':'3d'}, figsize=(12,6))
ax1 = axes[0]
ax1.plot_surface(x, y, f, cmap='plasma')
ax1.set_axis_off() 
ax1.view_init(elev=15, azim=45)

ax1 = axes[1]
ax1.plot_surface(x, y, f, cmap='plasma')
ax1.set_axis_off() 
ax1.view_init(elev=45, azim=45)
```

```python
_ = np.linspace(-3, 3, 100)
x, y = np.meshgrid(_, _)
f = np.exp(-y**2) * np.cos(x)
fig, ax =plt.subplots(subplot_kw={'projection':'3d'}, figsize=(120,6))
ax.plot_surface(x, y, f, cmap='plasma')
ax.set_axis_off() 
ax.set_title('3D Plot of $f(x, y) = e^{-(x^2 + y^2)}\\cos(x)$', fontsize=14)
```

```python
_ = np.linspace(-3.0, 3.0, 300)
x, y = np.meshgrid(_, _)
ratio = np.abs(np.cos(y)/np.sin(x))

with np.errstate(divide='ignore', invalid='ignore'):
    f = np.log(ratio)

f = np.clip(f, -5, 5)
fig, ax =plt.subplots(subplot_kw={'projection':'3d'}, figsize=(120,6))
ax.plot_surface(x, y, f, cmap='plasma')
ax.set_axis_off() 
ax.set_title('3D Plot of $f(x, y) = log\\left|\\frac{\\cos(y)}{\\sin(x)}\\right|$', fontsize=14)
ax.view_init(elev=45, azim=45)
```

```python
_ = np.linspace(-2, 2, 100)
x, y = np.meshgrid(_, _)
f = np.sin(x) * np.sin(y) * np.cos(x) * np.cos(y)
fig, ax =plt.subplots(subplot_kw={'projection':'3d'}, figsize=(6,6))
ax.plot_surface(x, y, f, cmap='plasma')
ax.set_axis_off() 
ax.set_title('3D Plot of $f(x, y) = sin(x).sin(y).cos(x).cos(y)$', fontsize=14)
```

```python
_ = np.linspace(-2, 2, 100)
x, y = np.meshgrid(_, _)
f = y * x**2 - y**3 
fig, ax =plt.subplots(subplot_kw={'projection':'3d'}, figsize=(6,6))
ax.plot_surface(x, y, f, cmap='plasma')
ax.set_axis_off() 
ax.set_title('3D Plot of $f(x, y) = x^2 y - y^3$', fontsize=14)
```
