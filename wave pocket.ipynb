Today, I dived a bit deeper into the equation I used on Day 3 code. 

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib import animation
from matplotlib.animation import PillowWriter
```

Defining the Function

```python
_ = np.linspace(-2, 2, 100)
x, y = np.meshgrid(_,_)

f = np.exp(-(x**2 + y**2))*np.sin(x)
```

The Plot I used earlier

```python
plt.contourf(x, y, f, levels=20, cmap='viridis')
plt.colorbar(label='Function Value')
plt.title('Contour Plot of $f(x, y) = e^{-(x^2 + y^2)}\\sin(x)$', fontsize=15)
plt.xlabel('X')
plt.ylabel('Y')
```

This is a much better type of contour graph, as it can tell you the value of each segment in a better way.

```python
cs = plt.contour(x, y, f, levels=20)
plt.clabel(cs, fontsize = 8)
```

Viewing the curve's real beauty.
We can clearly see two humps. One +ve and the other -ve.
It beautifully shows the concept of localised wave.

```python
fig, ax =plt.subplots(subplot_kw={'projection':'3d'}, figsize=(6,6))
ax.plot_surface(x, y, f, cmap='plasma')
ax.set_axis_off() 
ax.set_title('3D Plot of $f(x, y) = e^{-(x^2 + y^2)}\\sin(x)$', fontsize=14, pad=20)
```

If you runs the following code in a dedicated .py file, you'll get a beautiful animation. Which you'll definitely love. I'll add a gif file as \media\wave_pocket.gif
