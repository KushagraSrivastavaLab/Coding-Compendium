```python
_ = np.linspace(-1, 1, 100)
x, y = np.meshgrid(_,_)
z = x**2 + x*y
```

```python
plt.contourf(x, y, z, levels=30, vmin=0.02, cmap='plasma')
plt.colorbar(label="Voltaire")
```

```python
cs = plt.contour(x, y, z, levels=20)
plt.clabel(cs, fontsize = 8)
```

```python
fig, ax =plt.subplots(subplot_kw={'projection':'3d'})
ax.plot_surface(x, y, z)
```

```python
w = 3
_ = np.linspace(-3, 3, 100)
X, Y = np.meshgrid(_,_)
U = -1 - X**2 + Y
V = 1 + X - Y**2
speed = np.sqrt(U**2 + V**2)
```

```python
fig, axes = plt.subplots(2, 2, figsize=(7,7))
ax1 = axes[0][0]
ax1.streamplot(X, Y, U, V)
ax2 = axes[0][1]
ax2.streamplot(X, Y, U, V, color=speed)
ax3 = axes[1][0]
lw = 5 * speed / speed.max()
ax3.streamplot(X, Y, U, V, linewidth=lw)
ax4 = axes[1][1]
seedpoints = np.array([[0,1], [1,0], [1.2324, -1.2]])
ax4.streamplot(X, Y, U, V, start_points = seedpoints)
```
