```python
n = smp.symbols('n')
```

$\sum_{n=0}^{\infty} \frac{tan^{-1}{n}}{n^{1.1}}$ 

```python
smp.Sum(smp.atan(n)/n**smp.Rational(11,10), (n, 1, smp.oo)).doit()
#doit function couldn't find the answer as it tries to find a exact symbolic answer, which does not exists in this case
```

```python
smp.Sum(smp.atan(n)/n**smp.Rational(11,10), (n, 1, smp.oo)).n()
# n() is used to find the numeric value.
# It is really slow as it sums individually each element
```

$\sum_{n=1}^{\infty} \frac{1 + cos(n)}{n} $

```python
smp.Sum((1 + smp.cos(n))/n, (n, 1, smp.oo)).doit()
```

```python
smp.Sum((1 + smp.cos(n))/n, (n, 1, smp.oo)).n()
```

```python
# This answer is wrong. 
# n() can give wrong answers too.
# Why is it wrong?
# Because n() is used to find the numerical values of convergent functions, not divergent ones.
# This function grows really slowly in the starting, so the function thinks, that it is stoping to grow, so it gets truncated.
# As a result it just predicts a wrong value
```
